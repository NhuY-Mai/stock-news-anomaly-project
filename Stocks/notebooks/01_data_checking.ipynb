```python
import pandas as pd
import numpy as np
import os

# Đường dẫn tương đối từ thư mục gốc
FILE_PATH = r'D:\LUẬN VĂN\stock-news-anomaly-project\Stocks\data\raw\stocks_raw.csv'
OUTPUT_PATH = 'outputs/01_data_cleaned.csv'

os.makedirs('outputs', exist_ok=True)

df = pd.read_csv(FILE_PATH, sep=';')
print(f"Đã load {len(df)} dòng dữ liệu.")
```

```python
# 1. Ép kiểu ngày
df['date'] = pd.to_datetime(df['date'], format='%d/%m/%Y', errors='coerce')

# 2. Sắp xếp
df = df.sort_values(['ticker', 'date']).reset_index(drop=True)

# 3. Xóa trùng lặp (nếu có)
df = df.drop_duplicates(subset=['ticker', 'date'], keep='first')

# 4. Bỏ phiên không khớp lệnh (OHLCV = 0)
df = df[(df['volume'] > 0) & (df['open'] > 0) & (df['close'] > 0)].copy()

# 5. Tạo giá tham chiếu
df['ref_price'] = df.groupby('ticker')['close'].shift(1)
has_prior = df['prior_close_reference_vnd'].notna()
df.loc[has_prior, 'ref_price'] = df.loc[has_prior, 'prior_close_reference_vnd']

# Lưu file
df.to_csv(OUTPUT_PATH, index=False)
print(f"Đã lưu file sạch tại: {OUTPUT_PATH}")
```
